# Getting data into DataFrame
Importing pandas and display and the CSV file into a data frame and showing the first 5 entries ensuring the data was loaded


```python
from IPython.display import display
import pandas as pd
df = pd.read_csv('Electric_Vehicle_Population_Data.csv')
df.head()
```

# General Data
Next is the top 10 most popular counties, cities, states, makes, and model and shows the different types of Electric Vehicle Type(EVT) and the types of Clean Alternative Fuel Vehicle(CAFV) Eligibility and how many there are.
This is to show some top of the vehicles and types that are in the data set

```python
top_counties = df['County'].value_counts().head(10)
top_cities = df['City'].value_counts().head(10)
top_states = df['State'].value_counts().head(10)
top_makes = df['Make'].value_counts().head(10)
top_models = df['Model'].value_counts().head(10)

evt = df['Electric Vehicle Type'].value_counts().head(10)
cafv = df['Clean Alternative Fuel Vehicle (CAFV) Eligibility'].value_counts().head(10)

counties_df = top_counties.to_frame()
cities_df = top_cities.to_frame()
states_df = top_states.to_frame()
makes_df = top_makes.to_frame()
models_df = top_models.to_frame()
evt_df = evt.to_frame()
cafv_df = cafv.to_frame()



display(counties_df)
display(cities_df)
display(states_df)
display(makes_df)
display(models_df)
display(evt_df)
display(cafv_df)
```

## Data summary
This data is showing that most of the data is coming from Washington state with over 200,000 entries being from there and California being second on that list with 119


# Question 1: What is the Average Price for an EV

```python
avg_msrp = df[df['Base MSRP'] > 0]['Base MSRP'].mean()
print(f"Average MSRP of all EVs: ${avg_msrp:,.2f}")
```

### Question 1 Analysis:
Prices of vehicles vary greatly. Some are relatively inexpensive and only cost around \\$5,000 - \\$10,000, while other vehicles can cost upwards of \\$100,000 or more. The Price of luxury vehicles tend to cost more than their regular counterpart. EVs fall into a weird category since they tend to have futuristic features but have less usability and reliability for casual everyday drives or for long distance driving. With futuristic features but a lack of reliability the avg cost of EVs from this data is around $56,934.49

# Question 2: What is the average price of the EVs per make

```python
top_avg_vehicles = (
    df[['Make', 'Base MSRP']]
    .query('`Base MSRP` > 0')
    .groupby('Make', as_index=False)
    .agg({'Base MSRP': 'mean'})
    .sort_values(by='Base MSRP', ascending=False)
    .reset_index(drop=True)
)

top_avg_vehicles.index = top_avg_vehicles.index + 1
top_avg_vehicles.index.name = 'Rank'


display(top_avg_vehicles)
```

### Question 2 Analysis:
The cost of a vehicle can sway the opinion and feasibility for someone to buy it. The prices of gas vehicles can vary lots and EVs are no different. Porsche's and Fisker EVs are by far the most expensive at \\$135,551 and \\$102,000 respectively with Cadillac and Tesla being the next expensive at \\$75,095 and \\$70,216 with KIA being the lowest on the list at \\$32,276. With that price being very similar to that of a new gas car, this makes EVs a more tempting choice.

# Question 3: What are the year range of the EVs

# Question 4: What Are the Top 10 EVs With the Most Range
This question gets the make model and Electric Range, removes the entries with a range of 0 and averages out the ranges and ranks them from best to worst

```python
top_avg_vehicles = (
    df[['Make', 'Model', 'Electric Range']]
    .query('`Electric Range` > 0')
    .groupby(['Make', 'Model'], as_index=False)
    .agg({'Electric Range': 'mean'})
    .sort_values(by='Electric Range', ascending=False)
    .head(10)
    .reset_index(drop=True)
)

top_avg_vehicles.index = top_avg_vehicles.index + 1
top_avg_vehicles.index.name = 'Rank'


display(top_avg_vehicles)
```

### Question 4 Analysis:
One of the biggest draw backs of EVs are their range. gas vehicles can have a range of up to and beyond 500 miles while Evs tend to have much less. The highest range of an EV recorded on the dataset was 304 miles, which is still a long distance but not as long as most gas vehicles


# Question 5: What Are the Worst 10 EV Ranges
This question does the same as the previous but get the 10 worst instead of the best

```python
top_avg_vehicles = (
    df[['Make', 'Model', 'Electric Range', 'Clean Alternative Fuel Vehicle (CAFV) Eligibility']]
    .query('`Electric Range` > 0 and `Clean Alternative Fuel Vehicle (CAFV) Eligibility` == "Clean Alternative Fuel Vehicle Eligible"')
    .groupby(['Make', 'Model'], as_index=False)
    .agg({'Electric Range': 'mean'})
    .sort_values(by='Electric Range', ascending=False)
    .tail(10)
    .reset_index(drop=True)
)

top_avg_vehicles.index = top_avg_vehicles.index + 1
top_avg_vehicles.index.name = 'Rank'


display(top_avg_vehicles)

```

### Question 5 Analysis:
At the lowest end of the spectrum of range, we see that these EVs don't get much range at all.
From the data set the Electric Ranges go from 30 to 33 With the top 3 worst vehicles being:

1. Hyundai Santa Fe - 30
2. Cadillac CT6 - 31
3. BMW MX - 31


# Question 6: How many makes and models are there

# Question 7: What is the Average Range Per Make
This question will get the average Electric Range of each Make and shows the top 10 best range per Make

```python
top_avg_vehicles = (
    df[['Make', 'Electric Range']]
    .query('`Electric Range` > 0')
    .groupby('Make', as_index=False)
    .agg({'Electric Range': 'mean'})
    .sort_values(by='Electric Range', ascending=False)
    .head(10)
    .reset_index(drop=True)
)

top_avg_vehicles.index = top_avg_vehicles.index + 1
top_avg_vehicles.index.name = 'Rank'


display(top_avg_vehicles)
```

### Question 7 Analysis:
When buying a car, range is one of the most important factors since the longer the range the less you have to spend on fuel. With EVs generally having less range, it's important to know which EVs have enough range to support your daily activities. The table shows that Tesla, Jaguar, and Polestar have the highest Electric Range and have almost 100 more miles of range then the next choice.

# Question 8: What is the Average Range for each year

```python
range_by_year = (
    df[['Model Year', 'Electric Range']]
    .query('`Electric Range` > 0')
    .groupby('Model Year', as_index=False)
    .agg(
        Vehicle_Count=('Electric Range', 'count'),
        Electric_Range=('Electric Range', 'mean')
    )
    .sort_values(by='Model Year', ascending=False)
    .reset_index(drop=True)
)

range_by_year.index = range_by_year.index + 1
range_by_year.index.name = 'Rank'

display(range_by_year)
```

### Question 8 Analysis:
Over time product tend to get better and companies try to innovate. With range being one of the limiting factors in EVs, it makes sense that companies would try to innovate new ways to get their cars to travel for longer distances without having to charge. From the table we can see that there has been some improvement with range with more years having a higher range number
The tables shows that the average range of EVs have increased overtime with a bit of a bump from 2008-2010 and a dip from 2021-present.
